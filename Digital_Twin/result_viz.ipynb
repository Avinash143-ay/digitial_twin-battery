```python
import os
import numpy as np
import matplotlib.pyplot as plt

# Path to saved logs
log_dir = "logging_stats"

# Dictionary of all files to load
metrics = {
    "train_loss": "train_loss.npy",
    "train_v_mape": "train_v_mape.npy",
    "train_t_mape": "train_t_mape.npy",
    "train_max_v_mape": "train_max_v_mape.npy",
    "train_max_t_mape": "train_max_t_mape.npy",
    "train_v_confidence": "train_v_confidence.npy",
    "train_t_confidence": "train_t_confidence.npy",
    "train_vc_max": "train_vc_max.npy",
    "train_tc_max": "train_tc_max.npy",

    "val_loss": "val_loss.npy",
    "val_v_mape": "val_v_mape.npy",
    "val_t_mape": "val_t_mape.npy",
    "val_max_v_mape": "val_max_v_mape.npy",
    "val_max_t_mape": "val_max_t_mape.npy",
    "val_v_confidence": "val_v_confidence.npy",
    "val_t_confidence": "val_t_confidence.npy",
    "val_vc_max": "val_vc_max.npy",
    "val_tc_max": "val_tc_max.npy"
}

# Load all metrics into a dictionary
loaded_metrics = {}
for key, filename in metrics.items():
    filepath = os.path.join(log_dir, filename)
    if os.path.exists(filepath):
        loaded_metrics[key] = np.load(filepath)
    else:
        print(f"Warning: {filename} not found!")

# Example: Plot training vs validation loss
plt.figure(figsize=(8, 5))
plt.plot(loaded_metrics["train_loss"], label="Train Loss")
plt.plot(loaded_metrics["val_loss"], label="Validation Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Training & Validation Loss")
plt.legend()
plt.grid(True)
plt.show()

# Example: Plot v_mape and t_mape (train vs val)
plt.figure(figsize=(8, 5))
plt.plot(loaded_metrics["train_v_mape"], label="Train v_mape")
plt.plot(loaded_metrics["val_v_mape"], label="Val v_mape")
plt.plot(loaded_metrics["train_t_mape"], label="Train t_mape")
plt.plot(loaded_metrics["val_t_mape"], label="Val t_mape")
plt.xlabel("Epoch")
plt.ylabel("MAPE")
plt.title("MAPE Over Epochs")
plt.legend()
plt.grid(True)
plt.show()

# You can loop through metrics to plot them all if needed:
for key, values in loaded_metrics.items():
    plt.figure()
    plt.plot(values)
    plt.title(key)
    plt.xlabel("Epoch")
    plt.ylabel(key)
    plt.grid(True)
    plt.show()
```
