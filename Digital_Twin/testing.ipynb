```python

# Example sequence
current_seq_2 = torch.tensor([1.0, 3.0, 6.0])

# Number of interpolation points between original samples
upsample_factor = 2  # insert 1 interpolated value between each original pair

# Original positions
x_orig = torch.arange(len(current_seq_2))

# New positions
x_new = torch.linspace(0, len(current_seq_2) - 1, len(current_seq_2) * upsample_factor - (upsample_factor - 1))

# Interpolate linearly
current_seq_2_interp = torch.interp(x_new, x_orig, current_seq_2)

print(current_seq_2_interp)
```

```python
import numpy as np
import pandas as pd

# Step 1: Create a numpy array of size 6
curr_numpy = np.array([10, 20, 30, 40, 50, 60])
print("curr_numpy:", curr_numpy)

# Step 2: Create array with NaNs interleaved
curr_interp = np.zeros(2 * len(curr_numpy) - 1)
curr_interp[::2] = curr_numpy
curr_interp[1::2] = np.nan
print("curr_interp with NaNs:", curr_interp)

# Step 3: Interpolate NaNs
curr_interp = pd.Series(curr_interp).interpolate().values
print("curr_interp after interpolation:", curr_interp)
```
